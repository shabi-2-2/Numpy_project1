```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image1 = np.zeros((200,200,3),dtype=np.uint8)
# image1[10:60,20:90,0]
print("image shape:",image1.shape)
print("image dimensions:",image1.ndim)
print("image size:",image1.size)
```

```python
image1[10:60,20:90,:] = np.full([50,70,3],[255,0,0],dtype=np.uint8)
image1[10:60,110:180,:] = np.full([50,70,3],[0,255,0],dtype = np.uint8)
image1[80:140,20:90,:] = [0,0,255]
image1[80:140,110:180,:] = [255,255,0]
image1[155:175,:,:] = [0,255,255]

plt.figure(figsize=(6,6))
plt.imshow(image1)
plt.show()
```
