```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image1 = np.zeros((200,200,3),dtype=np.uint8)
# image1[10:60,20:90,0]
print("image shape:",image1.shape)
print("image dimensions:",image1.ndim)
print("image size:",image1.size)
```

```python
image1[10:60,20:90,:] = np.full([50,70,3],[255,0,0],dtype=np.uint8)
image1[10:60,110:180,:] = np.full([50,70,3],[0,255,0],dtype = np.uint8)
image1[80:140,20:90,:] = [0,0,255]
image1[80:140,110:180,:] = [255,255,0]
image1[155:175,:,:] = [0,255,255]

plt.figure(figsize=(6,6))
plt.imshow(image1)
plt.show()
```

```python
crop = image1[10:140,20:180,:]
flipped = image1[:,::-1,:]
red_channel = image1[:,:,0]
blue_channel = image1[:,:,1]
green_channel = image1[:,:,2]
print("red channel shape",red_channel.shape)
print("blue channel shape",blue_channel.shape)
print("green channel shape",green_channel.shape)
```

```python
fig ,axes = plt.subplots(2,3,figsize=(12,7))
axes[0,0].imshow(image1)
axes[0,0].set_title('original')

axes[0,1].imshow(crop)
axes[0,1].set_title('cropped')

axes[0,2].imshow(flipped)
axes[0,2].set_title('flipped')

axes[1,0].imshow(red_channel,cmap='grey',vmin=0,vmax=255)
axes[1,0].set_title('red channel')

axes[1,1].imshow(green_channel ,cmap='grey',vmin=0,vmax=255)
axes[1,1].set_title('green channel')

axes[1,2].imshow(blue_channel,cmap='grey',vmin=0,vmax=255)
axes[1,2].set_title('blue channel')


for ax in axes.flat:
  ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
# brightness and boolean masking

# increasing the brightness by 50
bright_float = image1.astype(np.float32) + 50
brightened = np.clip(bright_float ,0,255).astype(np.uint8)

# decreasing the brightness by 0.5(multiplying it by)
dark_float = image1.astype(np.float32)*0.5
darkened = np.clip(dark_float,0,255).astype(np.uint8)

# creating a brightness mask to check which pixels are too bright(>200)
bright_mask = np.max(image1,axis=2)>200

print('bright pixels',np.nonzero(bright_mask))
print('mask shape',bright_mask.shape)

# plotting all the things

fig,axes =plt.subplots(1,4,figsize=(14,4))

axes[0].imshow(image1)
axes[0].set_title('original')

axes[1].imshow(brightened)
axes[1].set_title('brightened')

axes[2].imshow(darkened)
axes[2].set_title('darkened')

axes[3].imshow(bright_mask,cmap='grey')
axes[3].set_title('bright pixels mask')

for ax in axes:
  ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
# comparing the memory usage
image1_uint8 = image1.astype(np.uint8)
image1_float32 = image1.astype(np.float32)
image1_float64 = image1.astype(np.float64)

# displaying the bytes
print(f'unit8 memory: {image1_uint8.nbytes} bytes')
print(f'float32 memory: {image1_float32.nbytes} bytes')
print(f'float64 memory: {image1_float64.nbytes} bytes')

print(f'memory ratios\n')
print(f'float32 / uint8 = {image1_float32.nbytes/image1_uint8.nbytes}')
print(f'float64 / unit8 = {image1_float64.nbytes/image1_uint8.nbytes}')

# final synthetic image
plt.figure(figsize=(5,5))
plt.imshow(image1_uint8)
plt.title('Final Synthetic image')
plt.axis('off')
plt.show()
```
